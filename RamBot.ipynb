```python
from langchain_community.document_loaders import TextLoader
loadder = TextLoader('data.txt', encoding="utf-8")
loadder
```

```python
text_documents = loadder.load()
text_documents
```

```python
#Text splitter
from langchain_text_splitters import RecursiveCharacterTextSplitter
textSplitter=RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=20)
finalDocuments=textSplitter.split_documents(text_documents)
finalDocuments
```

```python
texts = [doc.page_content for doc in finalDocuments]
```

```python
#Huggingface embedding
import os 
from dotenv import load_dotenv
load_dotenv
```

```python
os.environ['HF_Token']= os.getenv('HF_Token')
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings= HuggingFaceEmbeddings(model_name= "sentence-transformers/all-MiniLM-L6-v2")
```

```python
print(embeddings)
```

```python
#vector Store in DB
from langchain.community.vectorstores import FAISS
db= FAISS.from_documents(finalDocuments, embeddings)
```

```python
query= input(Enter your question : "What does the fabric filer do")
docs= db.similarity_search(query)
for i in range (len(docs)):
print(docs[i].page_content)
```
